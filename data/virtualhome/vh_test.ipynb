```python
import json
import random

# 读取原始数据
input_path = "/code/EUV/data/virtualhome/new_unseen_test.jsonl"
output_path = "/code/EUV/data/virtualhome/new_unseen_test_shuffled.jsonl"

with open(input_path, "r", encoding="utf-8") as f:
    lines = f.readlines()
    data = [json.loads(line) for line in lines]

# 打乱数据顺序
random.shuffle(data)

# 保存到新文件
with open(output_path, "w", encoding="utf-8") as f:
    for item in data:
        f.write(json.dumps(item, ensure_ascii=False) + "\n")

print(f"Shuffled data saved to {output_path}")
```
